```python
accepted_rows["supplier_name"] = accepted_rows["supplier_name"].str.strip()
```

```python
accepted_rows["missing_supplier_flag"] = accepted_rows["supplier_name"].isna().astype(int)
```

```python
accepted_rows["supplier_name"] = accepted_rows["supplier_name"].fillna("Not recorded")
```

```python
accepted_rows["missing_supplier_flag"].sum()
```

```python
recorded_status = ["Paid", "Approved", "Pending", "Under Review"]
```

```python
accepted_rows["Is_Recorded"] = accepted_rows["payment_status"].isin(recorded_status).astype(int)
```

```python
(accepted_rows["Is_Recorded"] == 0).sum()
```

```python
accepted_rows["Variance_GBP"] = accepted_rows["actual_spend_gbp"] - accepted_rows["budget_amount_gbp"]
```

```python
safe_budget = accepted_rows["budget_amount_gbp"].replace(0, float("nan"))
```

```python
accepted_rows["Variance_Pct"] = accepted_rows["Variance_GBP"] / safe_budget
```

```python
accepted_rows[["actual_spend_gbp", "budget_amount_gbp", "Variance_GBP", "Variance_Pct"]].head()
```

```python
accepted_rows["Recorded_Spend"] = accepted_rows["actual_spend_gbp"] * accepted_rows["Is_Recorded"]
supplier_total = accepted_rows.groupby(["council_name", "supplier_name"])["Recorded_Spend"].transform("sum")
council_total = accepted_rows.groupby("council_name")["Recorded_Spend"].transform("sum")
accepted_rows["Supplier_Share_Of_Council_Pct"] = supplier_total / council_total.replace(0, float("nan"))
```

```python
checks = {
    "missing_transaction_id": df["transaction_id"].isna().sum(),
    "duplicate_transaction_id": df["transaction_id"].duplicated().sum(),
    "missing_council_or_department": (df["council_name"].isna() | df["department"].isna()).sum(),
    "non_gbp_currency": (df["currency"].str.strip() != "GBP").sum(),
    "non_numeric_spend": pd.to_numeric(df["actual_spend_gbp"], errors="coerce").isna().sum(),
}
for name, count in checks.items():
    print(f"{name}: {count}")
assert sum(checks.values()) == 0, "New reject reasons found - add them to the rejected file before loading"
```

```python
summary = pd.DataFrame({
    "metric": ["source_rows", "accepted_rows", "rejected_INVALID_DATE", "review_NEGATIVE_SPEND", "missing_supplier_rows", "reconciled"],
    "value": [rows, len(accepted_rows), len(bad_date_rows), len(review_rows),
              accepted_rows["missing_supplier_flag"].sum(),
              len(accepted_rows) + len(bad_date_rows) + len(review_rows) == rows]
})
summary.to_csv(base + r"\processed\quality_summary.csv", index=False)
accepted_rows.to_csv(base + r"\processed\clean_council_spend.csv", index=False)
print(summary)
print(accepted_rows.shape)
```
